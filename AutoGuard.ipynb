```python

# Predictive Maintenance & Machine Failure Prediction
# ============================================================

import os
import sys
import zipfile
import urllib.request
import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import joblib

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.dummy import DummyClassifier
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    f1_score, roc_auc_score, average_precision_score,
    confusion_matrix, classification_report,
    precision_recall_curve, roc_curve
)
from sklearn.inspection import permutation_importance

sns.set_theme(style="whitegrid")
os.makedirs("autoguard_outputs", exist_ok=True)
os.makedirs("autoguard_models", exist_ok=True)

print("=" * 65)
print("AUTOGUARD AI | PREDICTIVE MAINTENANCE SYSTEM")
print("=" * 65)

# 1. DOWNLOAD DATASET

url = "https://archive.ics.uci.edu/static/public/601/ai4i+2020+predictive+maintenance+dataset.zip"

zip_path = "ai4i_dataset.zip"

urllib.request.urlretrieve(url, zip_path)

with zipfile.ZipFile(zip_path, "r") as z:
    csv_file = next(
        name for name in z.namelist()
        if name.endswith("ai4i2020.csv")
    )
    with z.open(csv_file) as f:
        df = pd.read_csv(f)

print("\nDataset downloaded successfully!")
print("Dataset shape:", df.shape)

# 2. DATA UNDERSTANDING

print("\nDATASET PREVIEW")
display(df.head())

print("\nDATASET INFORMATION")
df.info()

print("\nMISSING VALUES")
display(df.isnull().sum())

print("\nSTATISTICAL SUMMARY")
display(df.describe())

print("\nDUPLICATE ROWS:", df.duplicated().sum())

print("\nFAILURE DISTRIBUTION")
display(df["Machine failure"].value_counts())

# 3. EXPLORATORY DATA ANALYSIS

plt.figure(figsize=(7, 5))
sns.countplot(data=df, x="Machine failure")
plt.title("Machine Failure Distribution")
plt.xlabel("Machine Failure (0 = No, 1 = Yes)")
plt.ylabel("Number of Records")
plt.savefig("autoguard_outputs/failure_distribution.png")
plt.show()

plt.figure(figsize=(8, 5))
sns.countplot(data=df, x="Type", hue="Machine failure")
plt.title("Failure Distribution by Product Type")
plt.savefig("autoguard_outputs/failure_by_type.png")
plt.show()

sensor_cols = [
    "Air temperature [K]",
    "Process temperature [K]",
    "Rotational speed [rpm]",
    "Torque [Nm]",
    "Tool wear [min]"
]

df[sensor_cols].hist(figsize=(13, 9), bins=30)
plt.suptitle("Sensor Feature Distributions")
plt.tight_layout()
plt.savefig("autoguard_outputs/sensor_distributions.png")
plt.show()

plt.figure(figsize=(10, 7))
sns.heatmap(
    df[sensor_cols + ["Machine failure"]].corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)
plt.title("Sensor Correlation Heatmap")
plt.tight_layout()
plt.savefig("autoguard_outputs/correlation_heatmap.png")
plt.show()

# 4. FEATURE ENGINEERING

df["Temperature Difference [K]"] = (
    df["Process temperature [K]"] -
    df["Air temperature [K]"]
)

df["Mechanical Power [W]"] = (
    df["Torque [Nm]"] *
    df["Rotational speed [rpm]"] *
    (2 * np.pi / 60)
)

df["Wear Torque Interaction"] = (
    df["Tool wear [min]"] *
    df["Torque [Nm]"]
)

df["Temperature Ratio"] = (
    df["Process temperature [K]"] /
    df["Air temperature [K]"]
)

print("\nFEATURE ENGINEERING COMPLETED")

# 5. PREPARE MACHINE LEARNING DATA

target = "Machine failure"

# Remove identifiers and all failure-mode labels.
# TWF, HDF, PWF, OSF and RNF would reveal the failure outcome.

drop_cols = [
    "UDI",
    "Product ID",
    "TWF",
    "HDF",
    "PWF",
    "OSF",
    "RNF",
    target
]

X = df.drop(columns=drop_cols)
y = df[target]

categorical_features = ["Type"]
numerical_features = [
    col for col in X.columns
    if col not in categorical_features
]

preprocessor = ColumnTransformer(
    transformers=[
        (
            "numeric",
            Pipeline([
                ("imputer", SimpleImputer(strategy="median")),
                ("scaler", StandardScaler())
            ]),
            numerical_features
        ),
        (
            "categorical",
            Pipeline([
                ("imputer", SimpleImputer(strategy="most_frequent")),
                ("encoder", OneHotEncoder(handle_unknown="ignore"))
            ]),
            categorical_features
        )
    ]
)

# 6. TRAIN / VALIDATION / TEST SPLIT

X_train_full, X_test, y_train_full, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

X_train, X_val, y_train, y_val = train_test_split(
    X_train_full,
    y_train_full,
    test_size=0.20,
    random_state=42,
    stratify=y_train_full
)

print("\nTRAINING RECORDS:", len(X_train))
print("VALIDATION RECORDS:", len(X_val))
print("TEST RECORDS:", len(X_test))

# 7. MODEL TRAINING

models = {
    "Logistic Regression": LogisticRegression(
        max_iter=2000,
        class_weight="balanced",
        random_state=42
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=300,
        max_depth=12,
        min_samples_split=5,
        class_weight="balanced_subsample",
        random_state=42,
        n_jobs=-1
    )
}

fitted_models = {}
validation_results = []

for name, model in models.items():

    pipeline = Pipeline([
        ("preprocessor", preprocessor),
        ("classifier", model)
    ])

    pipeline.fit(X_train, y_train)

    probabilities = pipeline.predict_proba(X_val)[:, 1]

    precision, recall, thresholds = precision_recall_curve(
        y_val, probabilities
    )

    f1_values = (
        2 * precision[:-1] * recall[:-1] /
        (precision[:-1] + recall[:-1] + 1e-10)
    )

    best_index = np.argmax(f1_values)
    best_threshold = thresholds[best_index]

    predictions = (probabilities >= best_threshold).astype(int)

    fitted_models[name] = {
        "pipeline": pipeline,
        "threshold": best_threshold
    }

    validation_results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_val, predictions),
        "Precision": precision_score(y_val, predictions, zero_division=0),
        "Recall": recall_score(y_val, predictions, zero_division=0),
        "F1 Score": f1_score(y_val, predictions, zero_division=0),
        "PR AUC": average_precision_score(y_val, probabilities),
        "Threshold": best_threshold
    })

validation_df = pd.DataFrame(validation_results)

print("\nVALIDATION MODEL COMPARISON")
display(validation_df.round(4))

# Select model using validation F1 score
best_model_name = validation_df.sort_values(
    "F1 Score", ascending=False
).iloc[0]["Model"]

best_pipeline = fitted_models[best_model_name]["pipeline"]
best_threshold = fitted_models[best_model_name]["threshold"]

print("\nSELECTED MODEL:", best_model_name)
print("VALIDATION-SELECTED THRESHOLD:", round(best_threshold, 4))

# 8. FINAL TEST EVALUATION

test_probabilities = best_pipeline.predict_proba(X_test)[:, 1]

test_predictions = (
    test_probabilities >= best_threshold
).astype(int)

print("\nFINAL TEST RESULTS")
print("=" * 50)

metrics = {
    "Accuracy": accuracy_score(y_test, test_predictions),
    "Precision": precision_score(y_test, test_predictions, zero_division=0),
    "Recall": recall_score(y_test, test_predictions, zero_division=0),
    "F1 Score": f1_score(y_test, test_predictions, zero_division=0),
    "ROC AUC": roc_auc_score(y_test, test_probabilities),
    "PR AUC": average_precision_score(y_test, test_probabilities)
}

for metric, value in metrics.items():
    print(f"{metric}: {value:.4f}")

print("\nCLASSIFICATION REPORT")
print(classification_report(
    y_test,
    test_predictions,
    target_names=["Normal", "Failure"],
    zero_division=0
))

# 9. CONFUSION MATRIX

cm = confusion_matrix(y_test, test_predictions)

plt.figure(figsize=(7, 5))
sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["Normal", "Failure"],
    yticklabels=["Normal", "Failure"]
)
plt.title("AutoGuard AI - Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.tight_layout()
plt.savefig("autoguard_outputs/confusion_matrix.png")
plt.show()

# 10. ROC CURVE

fpr, tpr, _ = roc_curve(y_test, test_probabilities)

plt.figure(figsize=(7, 5))
plt.plot(fpr, tpr, label=f"ROC AUC = {metrics['ROC AUC']:.3f}")
plt.plot([0, 1], [0, 1], linestyle="--")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve")
plt.legend()
plt.savefig("autoguard_outputs/roc_curve.png")
plt.show()

# 11. PRECISION-RECALL CURVE

precision, recall, _ = precision_recall_curve(
    y_test, test_probabilities
)

plt.figure(figsize=(7, 5))
plt.plot(recall, precision)
plt.xlabel("Recall")
plt.ylabel("Precision")
plt.title("Precision-Recall Curve")
plt.savefig("autoguard_outputs/precision_recall_curve.png")
plt.show()

# 12. FEATURE IMPORTANCE

importance_result = permutation_importance(
    best_pipeline,
    X_test,
    y_test,
    scoring="average_precision",
    n_repeats=5,
    random_state=42,
    n_jobs=-1
)

importance_df = pd.DataFrame({
    "Feature": X.columns,
    "Importance": importance_result.importances_mean
}).sort_values("Importance", ascending=False)

print("\nFEATURE IMPORTANCE")
display(importance_df.round(5))

plt.figure(figsize=(10, 6))
sns.barplot(
    data=importance_df.head(10),
    x="Importance",
    y="Feature"
)
plt.title("Top 10 Features Affecting Predictions")
plt.tight_layout()
plt.savefig("autoguard_outputs/feature_importance.png")
plt.show()

# 13. TEST PREDICTION TABLE

prediction_results = X_test.copy()

prediction_results["Actual Failure"] = y_test.values
prediction_results["Predicted Failure"] = test_predictions
prediction_results["Failure Probability"] = test_probabilities

prediction_results["Prediction Status"] = np.where(
    test_predictions == 1,
    "POTENTIAL FAILURE",
    "NORMAL"
)

print("\nSAMPLE PREDICTIONS")
display(prediction_results.head(15))

prediction_results.to_csv(
    "autoguard_outputs/test_predictions.csv",
    index=False
)

# 14. CUSTOM MACHINE FAILURE PREDICTION

def predict_machine_failure(
    product_type="M",
    air_temperature=300.0,
    process_temperature=310.0,
    rotational_speed=1500,
    torque=40.0,
    tool_wear=100
):

    temperature_difference = (
        process_temperature - air_temperature
    )

    mechanical_power = (
        torque * rotational_speed * (2 * np.pi / 60)
    )

    wear_torque = tool_wear * torque

    temperature_ratio = (
        process_temperature / air_temperature
    )

    input_data = pd.DataFrame([{
        "Type": product_type,
        "Air temperature [K]": air_temperature,
        "Process temperature [K]": process_temperature,
        "Rotational speed [rpm]": rotational_speed,
        "Torque [Nm]": torque,
        "Tool wear [min]": tool_wear,
        "Temperature Difference [K]": temperature_difference,
        "Mechanical Power [W]": mechanical_power,
        "Wear Torque Interaction": wear_torque,
        "Temperature Ratio": temperature_ratio
    }])

    probability = best_pipeline.predict_proba(
        input_data
    )[0][1]

    prediction = int(probability >= best_threshold)

    print("\nAUTOGUARD AI DIAGNOSTIC REPORT")
    print("-" * 45)
    print("Product Type:", product_type)
    print("Failure Probability:", f"{probability * 100:.2f}%")

    if prediction == 1:
        print("STATUS: POTENTIAL FAILURE DETECTED")
        print("Recommendation: Inspect machine and schedule maintenance.")
    else:
        print("STATUS: NORMAL OPERATING CONDITION")
        print("Recommendation: Continue routine monitoring.")

    print("-" * 45)

    return {
        "prediction": prediction,
        "failure_probability": float(probability),
        "threshold": float(best_threshold)
    }

# Try a custom machine
sample_prediction = predict_machine_failure(
    product_type="M",
    air_temperature=300.0,
    process_temperature=310.0,
    rotational_speed=1500,
    torque=40.0,
    tool_wear=180
)

# 15. SAVE MODEL AND RESULTS

joblib.dump(
    best_pipeline,
    "autoguard_models/autoguard_model.pkl"
)

joblib.dump(
    {
        "threshold": best_threshold,
        "model_name": best_model_name,
        "features": list(X.columns)
    },
    "autoguard_models/model_metadata.pkl"
)

validation_df.to_csv(
    "autoguard_outputs/model_comparison.csv",
    index=False
)

pd.DataFrame([metrics]).to_csv(
    "autoguard_outputs/final_metrics.csv",
    index=False
)

# 16. PROJECT SUMMARY

print("\n" + "=" * 65)
print("AUTOGUARD AI PROJECT COMPLETED")
print("=" * 65)

print("Dataset rows:", len(df))
print("Features used:", X.shape[1])
print("Best model:", best_model_name)
print("Test accuracy:", round(metrics["Accuracy"], 4))
print("Test precision:", round(metrics["Precision"], 4))
print("Test recall:", round(metrics["Recall"], 4))
print("Test F1:", round(metrics["F1 Score"], 4))
print("\nSaved model: autoguard_models/autoguard_model.pkl")
print("Saved outputs: autoguard_outputs/")
print("\nAll project outputs are ready.")

# 17. CREATE DOWNLOADABLE PROJECT ZIP

import shutil

shutil.make_archive(
    "AutoGuard_AI_Day10",
    "zip",
    ".",
    base_dir="autoguard_outputs"
)

# Add trained model files to a separate archive
shutil.make_archive(
    "AutoGuard_AI_Models",
    "zip",
    ".",
    base_dir="autoguard_models"
)

print("\nDOWNLOAD FILES:")
print("AutoGuard_Day10.zip")
print("AutoGuard_Models.zip")
```

```python

# ============================================================
# AUTOGUARD AI - INTERACTIVE PREDICTION DASHBOARD
# ============================================================

import ipywidgets as widgets
from IPython.display import display, clear_output

# Load saved model
loaded_model = joblib.load(
    "autoguard_models/autoguard_model.pkl"
)

metadata = joblib.load(
    "autoguard_models/model_metadata.pkl"
)

threshold = metadata["threshold"]

# Input controls
product_type = widgets.Dropdown(
    options=["L", "M", "H"],
    value="M",
    description="Type:"
)

air_temp = widgets.FloatSlider(
    value=300,
    min=295,
    max=305,
    step=0.5,
    description="Air Temp:"
)

process_temp = widgets.FloatSlider(
    value=310,
    min=305,
    max=315,
    step=0.5,
    description="Process Temp:"
)

rpm = widgets.IntSlider(
    value=1500,
    min=1000,
    max=2500,
    step=50,
    description="RPM:"
)

torque = widgets.FloatSlider(
    value=40,
    min=10,
    max=80,
    step=1,
    description="Torque:"
)

wear = widgets.IntSlider(
    value=100,
    min=0,
    max=250,
    step=5,
    description="Tool Wear:"
)

button = widgets.Button(
    description="ANALYZE MACHINE",
    button_style="primary",
    icon="search"
)

output = widgets.Output()

def analyze_machine(b):

    with output:
        clear_output()

        temp_diff = process_temp.value - air_temp.value

        mechanical_power = (
            torque.value * rpm.value * (2 * np.pi / 60)
        )

        wear_torque = wear.value * torque.value

        temp_ratio = process_temp.value / air_temp.value

        input_data = pd.DataFrame([{
            "Type": product_type.value,
            "Air temperature [K]": air_temp.value,
            "Process temperature [K]": process_temp.value,
            "Rotational speed [rpm]": rpm.value,
            "Torque [Nm]": torque.value,
            "Tool wear [min]": wear.value,
            "Temperature Difference [K]": temp_diff,
            "Mechanical Power [W]": mechanical_power,
            "Wear Torque Interaction": wear_torque,
            "Temperature Ratio": temp_ratio
        }])

        probability = loaded_model.predict_proba(
            input_data
        )[0][1]

        prediction = int(probability >= threshold)

        print("=" * 45)
        print("       AUTOGUARD AI DIAGNOSTIC")
        print("=" * 45)

        print(f"\nFailure Probability: {probability*100:.2f}%")
        print(f"Decision Threshold: {threshold:.3f}")

        if prediction == 1:
            print("\nSTATUS: POTENTIAL FAILURE")
            print("Suggested Action: Schedule a machine inspection.")
        else:
            print("\nSTATUS: NORMAL")
            print("Suggested Action: Continue routine monitoring.")

        print("\nNOTE: This is a model prediction, not a")
        print("confirmed engineering safety assessment.")

        plt.figure(figsize=(7, 2))
        plt.barh(
            ["Predicted Failure Probability"],
            [probability],
            color="tomato" if prediction else "seagreen"
        )
        plt.xlim(0, 1)
        plt.xlabel("Probability")
        plt.title("AutoGuard AI Risk Score")
        plt.show()

button.on_click(analyze_machine)

display(widgets.VBox([
    widgets.HTML("<h2>AutoGuard AI | Machine Diagnostics</h2>"),
    product_type,
    air_temp,
    process_temp,
    rpm,
    torque,
    wear,
    button,
    output
]))
```
